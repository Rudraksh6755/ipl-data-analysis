# Indexing and Selection

We also look at `idxmax()` and `idxmin()` to locate rows holding the maximum and minimum values of a column.

## Indexing Operator `[]`

The `[]` operator is the simplest way to access data in a DataFrame. It is mainly used for column selection.

**Selecting a Single Column:** passing a single column name as a string returns a Series.

````python
df['Batter']
````

**Selecting Multiple Columns:** passing a list of column names returns a DataFrame.

````python
df[['Batter', 'Team', 'Method']]
````

While `[]` is convenient for selecting columns, it lacks row selection and customized index handling. This is where `.loc[]` and `.iloc[]` come in.

## `.loc[]` - Label-based Selection

`.loc[]` selects data by labels (row/column names). It supports slicing and boolean conditions, and can handle numerical indices when they are set as row labels.

**Syntax:**
```python
df.loc[row_label, column_label]
```

**Single Row Selection:** select a single row by its index label.

````python
df.loc[1178406]
````

````python
df.loc[[548315]]
````

**Multiple Row Selection:** select multiple rows by their index labels.

````python
df.loc[[829795, 1178406]]
````

**Single Column Selection:** select one column across all rows.

````python
df.loc[:, 'Team']
````

**Multiple Columns Selection:** select multiple columns across all rows.

````python
df.loc[:, ['Batter', 'Team']]
````

**Row Slicing:** select a range of rows using labels (inclusive of the end label).

````python
df.loc[829795:733975]
````

**Column Slicing:** select a range of columns using column labels.

````python
df.loc[:, 'Batter':'Venue']
````

**Conditional Row Selection:** select rows that satisfy a given condition.

````python
df.loc[df['Method'] == 'Bowled']
````

````python
df.loc[df['Method'] != 'lbw']
````

**Row and Column Selection:** select specific rows and specific columns together.

````python
df.loc[[829779, 1175372], 'Batter':'Venue']
````

## `.iloc[]` - Position-based Indexing

`.iloc[]` selects data by integer index positions. Rows and columns are selected based on their numerical position in the DataFrame, using zero-based indexing.

**Syntax:**
```python
df.iloc[row_index, column_index]
```

**Single Row Selection:** select a single row by its position.

````python
df.iloc[2]
````

**Multiple Rows Selection:** select multiple rows by their positions.

````python
df.iloc[[0, 2, 3]]
````

**Single Column Selection:** select one column by its position.

````python
df.iloc[:, 3]
````

**Multiple Columns Selection:** select multiple columns by their positions.

````python
df.iloc[:, [0, 3]]
````

**Row Slicing:** select a range of rows by position (end position excluded).

````python
df.iloc[1:3]
````

**Column Slicing:** select a range of columns by position.

````python
df.iloc[:, 1:3]
````

**Row and Column Selection:** select rows and columns together by position.

````python
df.iloc[1:3, 0:3]
````

## `.loc[]` vs `.iloc[]`

| Aspect | `.loc[]` | `.iloc[]` |
|---|---|---|
| Selection basis | Labels (row/column names) | Integer positions |
| Indexing style | Label-based | Zero-based positional |
| Slicing end | Inclusive of end label | Excludes end position |
| Conditions | Supports boolean conditions | Positional only |

## `idxmax()` and `idxmin()`

These return the index label of the maximum and minimum values in a Series, letting us locate the full row where an extreme value occurs. Replace `'ColumnName'` with a numeric column from your dataset.

**`idxmax()`:** returns the index label of the maximum value in a column, used to locate the row where the highest value occurs.

````python
df.loc[df['Team'].idxmax()]
````

**`idxmin()`:** returns the index label of the minimum value in a column, used to locate the row where the lowest value occurs.

````python
df.loc[df['Venue'].idxmin()]
````

## Conclusion

The `[]` operator is best for quick column selection. `.loc[]` selects by labels and supports slicing and boolean conditions, while `.iloc[]` selects by integer position. `idxmax()` and `idxmin()` combined with `.loc[]` let us retrieve the full row holding the maximum or minimum value of a column.
